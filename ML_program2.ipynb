```python
import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Weak'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

attributes = list(df.columns[:-1])
n = len(attributes)

S = ['0'] * n
G = [['?'] * n]

for _, row in df.iterrows():

    if row.iloc[-1] == 'Yes':

        # Remove inconsistent general hypotheses
        new_G = []

        for g in G:
            if all(g[i] == '?' or g[i] == row.iloc[i] for i in range(n)):
                new_G.append(g)

        G = new_G

        # Generalize S
        for i in range(n):
            if S[i] == '0':
                S[i] = row.iloc[i]
            elif S[i] != row.iloc[i]:
                S[i] = '?'

    else:
        # Specialize G
        new_G = []

        for g in G:
            for i in range(n):
                if g[i] == '?':
                    if S[i] != row.iloc[i] and S[i] != '0':
                        temp = g.copy()
                        temp[i] = S[i]
                        new_G.append(temp)

        if new_G:
            G = new_G

print("Specific Boundary S:")
print(S)

print("\nGeneral Boundary G:")
for g in G:
    print(g)
```
